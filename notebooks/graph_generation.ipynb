# Generación de Grafos con NetworkX

Este notebook genera y visualiza grafos de tipo **Erdős-Rényi** y **Barabási-Albert**.

```python
import networkx as nx
import matplotlib.pyplot as plt
```

## 1. Grafos Erdős-Rényi
El modelo Erdős-Rényi $G(n, p)$ recibe dos parámetros:
- $n$: número de nodos
- $p$: probabilidad de crear una arista entre cualquier par de nodos

A continuación, graficamos 4 redes variando $n$ y $p$.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 14))

er_params = [
    {'n': 30, 'p': 0.05},
    {'n': 30, 'p': 0.20},
    {'n': 100, 'p': 0.02},
    {'n': 100, 'p': 0.10}
]

for ax, params in zip(axes.flat, er_params):
    G = nx.erdos_renyi_graph(n=params['n'], p=params['p'])
    nx.draw_networkx(G, ax=ax, node_size=400, with_labels=True, node_color='blue', font_color='white', font_size=8, edge_color='gray')
    ax.set_title(f"Erdős-Rényi\n$n={params['n']}, p={params['p']}$")

plt.tight_layout()
plt.show()
```

## 2. Grafos Barabási-Albert
El modelo estándar de Barabási-Albert utiliza los parámetros $n$ (nodos) y $m$ (aristas conectadas en cada paso). 
Puesto que se solicita variar $n$, $m$ y una probabilidad $p$, usaremos la variante de **Holme y Kim** (`powerlaw_cluster_graph` en NetworkX). Este modelo es esencialmente el de Barabási-Albert, en el cual cada nueva arista aleatoria es seguida por una probabilidad $p$ de formar un triángulo (aumentando el clustering).

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 14))

ba_params = [
    {'n': 50, 'm': 2, 'p': 0.1},
    {'n': 50, 'm': 5, 'p': 0.8},
    {'n': 200, 'm': 2, 'p': 0.1},
    {'n': 200, 'm': 5, 'p': 0.8}
]

for ax, params in zip(axes.flat, ba_params):
    G = nx.powerlaw_cluster_graph(n=params['n'], m=params['m'], p=params['p'])
    nx.draw_networkx(G, ax=ax, node_size=400, with_labels=True, node_color='blue', font_color='white', font_size=8, edge_color='gray')
    ax.set_title(f"Barabási-Albert (variante con prob $p$)\n$n={params['n']}, m={params['m']}, p={params['p']}$")

plt.tight_layout()
plt.show()
```
